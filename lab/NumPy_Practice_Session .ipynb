**Q7.1** Set a seed, generate a 3x3 uniform random array, and confirm reproducibility by rerunning with the same seed.

```python
# Write your code here


np.random.seed(23)

arr1 = np.random.rand(3, 3)

print("First run:")
print(arr1)


np.random.seed(23)
arr2 = np.random.rand(3, 3)


print(arr2)


print("\nReproducible:", np.array_equal(arr1, arr2))
```

## 12. Stacking & Concatenation

**Q12.1** Create two 2x2 arrays. Combine them using vstack, hstack, and concatenate on both axis=0 and axis=1.

```python
# Write your code here
t1 = np.arange(4).reshape(2,2)
t2 = np.arange(4,8).reshape(2,2)
print(np.vstack((t1,t2)))
print(np.hstack((t1,t2)))
print(np.concatenate((t1,t2),axis=0))
print(np.concatenate((t1,t2),axis=1))
```

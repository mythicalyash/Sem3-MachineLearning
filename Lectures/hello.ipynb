```python
name = "  rahul KUmar Sharma"

def format_name(name):
    name = name.strip()
    name = name.title()
    return name
def validate_name(name):
    name = "".join(name.split())
    if name.isalpha():
        return True
    else:
        return False    
name = format_name(name)
print(validate_name(name))
print(name)


```

```python
s = "1 ,2, 3, 4 "
num = list(map(int, s.split(",")))
sum = 0;
avg = 0;

for i in num:
    sum+=i
    avg = sum/len(num)
print("Sum: ", sum)
print("Average: ", avg)
    
```

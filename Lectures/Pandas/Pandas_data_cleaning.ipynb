## 📌 Recap

### Finding
```python
p.isnull().sum()      # blanks per column   (isna() is the same function)
p.info()              # non-null counts and types
p.describe()          # min, median, max — where the extremes hide
```

### The three options

| Option | Code | Use it when |
|---|---|---|
| **Delete** | `p.dropna(subset=['mass'])` | few blanks, and you can spare the rows |
| **Fill** | `p['mass'].fillna(value)` | the blank is a genuine unknown |
| **Transform** | `p['mass'].notnull()` | the blank itself means something |

⚠️ `dropna()` with no `subset` deletes on **any** blank: 1035 → 498 here, 891 → 183 on Titanic.

### Filling honestly
- **median** for numbers with a long tail — the mean gets dragged by extremes
- **mode** for text — `t['Embarked'].mode()[0]`
- every fill narrows the spread and invents rows that never existed
- ⭐ ask **who** is missing before **what** to fill. On Titanic, 136 of the 177 missing ages
  were third class, so a single global fill hides that pattern instead of handling it.

### Duplicates and saving
```python
p.duplicated().sum()                 # exact copies only
p.drop_duplicates()
p.duplicated(subset=['method'])      # "same" is whatever you define it to be
p.to_csv('clean.csv', index=False)   # index=False, always
```

```python
import numpy as np
```

```python
#arithmatic operations
a = np.arange(10)
print(a)
a= a+5
print("added 5:", a)
print("multiplied by 5:", a*5)
print("exponentiated:", a**2)
```

```python
# now with 2 arrays

x = np.arange(10)
print(x)
y =np.arange(11,21)
print(y)
print("added:", x+y)
print("multiplied:", x*y)
print("dot product:", x@y)

```

```python
#comparision operations - boolean arrays

x = np.arange(5)
y = np.arange(1,6)
print("x:", x)
print("y:", y)
print(x==y)
print(x>y)
print(x<y)
```

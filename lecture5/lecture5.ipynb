```python
# stacking arrays

r = np.array([[1,2,3],[4,5,6]])
s = np.array([[7,8,9],[10,11,12]])
arr4 = np.arange(12).reshape(2,6)
print("row wise stacking:",np.vstack((r,s)))
print("column wise stacking:",np.hstack((r,s)))
```

```python
# concatenation of arrays
# np.concatenate(arrays, axis=0)
# axis=0 for row wise concatenation
# axis=1 for column wise concatenation
# jis order me array pass krenge usi order me stacking hogi 

print("row wise concatenation:", np.concatenate((r,s), axis=0)) # row wise concatenation
print("column wise concatenation:", np.concatenate((r,s), axis=1)) # column wise concatenation

```
